### Turftopic - CVP testng

```python
# small subset of data
calculator = data["interactions"].loc[data["interactions"]["app"] == "Calculator"]
calc_mis = calculator.loc[calculator["defender_prompt_strategy"] == "mislead"]
calc_vul = calculator.loc[calculator["defender_prompt_strategy"] == "vulnerable"]

# Drop empty strings
calc_att_mis_strats = (
    calc_mis["attacker_strategy"]
    .dropna()
    .loc[lambda s: s.str.strip().ne("")]
    .tolist()
)

calc_att_vul_strats = (
    calc_vul["attacker_strategy"]
    .dropna()
    .loc[lambda s: s.str.strip().ne("")]
    .tolist()
)

calc_def_mis_strats = (
    calc_mis["defender_strategy"]
    .dropna()
    .loc[lambda s: s.str.strip().ne("")]
    .tolist()
)

calc_def_vul_strats = (
    calc_vul["defender_strategy"]
    .dropna()
    .loc[lambda s: s.str.strip().ne("")]
    .tolist()
)

print(calc_att_mis_strats[1])
print()
print(calc_def_mis_strats[1])
print("\n"*3)
print(calc_att_vul_strats[1])
print()
print(calc_def_vul_strats[1])
```

```python
from turftopic import ConceptVectorProjection

# believable
positive = [
    "Highly secure",
    "Best practice",
    ]

# anti-believable = suspicious
negative = [
   "Vulnerable",
   "Suspicious",
]
cvp = ConceptVectorProjection(seeds=(positive, negative))

att_mis_doc_concept_matrix = cvp.transform(calc_att_mis_strats)
print(att_mis_doc_concept_matrix)

print("\n"*3)

att_vul_doc_concept_matrix = cvp.transform(calc_att_vul_strats)
print(att_vul_doc_concept_matrix)

print("\n"*3)

print("\nDefender: ")
def_mis_doc_concept_matrix = cvp.transform(calc_def_mis_strats)
print(def_mis_doc_concept_matrix)

print("\n"*3)

def_vul_doc_concept_matrix = cvp.transform(calc_def_vul_strats)
print(def_vul_doc_concept_matrix)
print("\n"*3)
```
