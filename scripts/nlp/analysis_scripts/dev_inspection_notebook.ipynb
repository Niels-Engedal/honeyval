## Calculate CVP for each statement line

```python
# Define placeholder seeds
seeds = {
    "believability": (
        # Positive: believable
        [
            "The response appears authentic and credible.",
            "The information is realistic and technically plausible.",
            "The details are consistent with the expected system behavior.",
            "The returned data looks genuine and trustworthy.",
            "The output makes sense in the context of this application."
        ],
        # Negative: unbelievable
        [
            "The response appears fabricated and misleading.",
            "The information is unrealistic and technically implausible.",
            "The details contradict the expected system behavior.",
            "The returned data looks fake and untrustworthy.",
            "The output makes no sense in the context of this application."
        ]
    ),
    "enticingness": (
        # Positive: enticing
        [
            "This response reveals a promising lead worth pursuing.",
            "The information offers an attractive opportunity.",
            "The output contains useful clues for achieving my objective.",
            "The result motivates further investigation.",
            "The details suggest an interesting and actionable next step."
        ],
        # Negative: unenticing
        [
            "This response provides no promising leads worth pursuing.",
            "The information offers no attractive opportunities.",
            "The output contains no useful clues for achieving my objective.",
            "The result discourages further investigation.",
            "The details suggest no interesting or actionable next steps."
        ]
    )
}
```

### Etablish CVP

```python
from sentence_transformers import SentenceTransformer

encoderModel = SentenceTransformer(
    "cisco-ai/SecureBERT2.0-biencoder"
)

cvp = ConceptVectorProjection(seeds=seeds, encoder=encoderModel)
```

```python
# Exclude missing and empty strategy text
valid = df_st["strategy"].map(
    lambda x: isinstance(x, str) and bool(x.strip())
)

# Compute both CVP scores at once
scores = cvp.transform(
    df_st.loc[valid, "strategy"].tolist()
)

# Add scores to original DataFrame
score_cols = ["cvp_believability", "cvp_enticingness"]

df_st[score_cols] = np.nan
df_st.loc[valid, score_cols] = scores

# Inspect results
df_st[
    ["actor", "event", "strategy",
     "cvp_believability", "cvp_enticingness"]
].head()
```

```python
import matplotlib.pyplot as plt

def plot_cvp_trajectory(df, interaction_id, actor="attacker"):

    subset = (
        df.loc[
            (df["interaction_id"] == interaction_id)
            & (df["actor"] == actor)
        ]
        .sort_values("statement_order")
    )

    if subset.empty:
        raise ValueError("No statements found for this selection.")

    # Plot raw scores and deltas
    for prefix, ylabel, title in [
        ("cvp_", "CVP score", "Raw CVP scores"),
        ("delta_", "Δ CVP score", "Changes in CVP scores")
    ]:
        fig, ax = plt.subplots(figsize=(11, 4))

        for concept in ["believability", "enticingness"]:
            ax.plot(
                subset["statement_order"],
                subset[f"{prefix}{concept}"],
                marker="o",
                label=concept.capitalize()
            )

        if prefix == "delta_":
            ax.axhline(0, color="gray", linestyle="--")

        ax.set(
            title=f"{title} — {actor}",
            xlabel="Statement order",
            ylabel=ylabel
        )
        ax.legend()
        ax.grid(alpha=0.2)

        plt.tight_layout()
        plt.show()
```

```python
# Example: the interaction with the most attacker statements
interaction_id = (
    df_st.loc[df_st["actor"] == "attacker", "interaction_id"]
    .value_counts()
    .idxmax()
)

plot_cvp_trajectory(df_st, interaction_id, actor="attacker")
plot_cvp_trajectory(df_st, interaction_id, actor="defender")
```

## Inspect attacker differences to defender prompt strategies

```python
# Select attacker action statements
df_cmp = df_st.loc[
    (df_st["actor"] == "attacker") &
    (df_st["event"] == "TOOL")
].copy()

# Attach experimental metadata if not already present
interactions = data["interactions"].set_index("interaction_id")

for col in ["defender_prompt_strategy", "app"]:
    if col not in df_cmp.columns:
        df_cmp[col] = df_cmp["interaction_id"].map(interactions[col])

# Keep the two experimental conditions
df_cmp = df_cmp.loc[
    df_cmp["defender_prompt_strategy"].isin(["vulnerable", "mislead"])
].copy()

# Keep statements with valid CVP scores
concepts = ["believability", "enticingness"]

df_cmp = df_cmp.dropna(
    subset=[f"cvp_{c}" for c in concepts]
)

# Sort chronologically
df_cmp = df_cmp.sort_values(
    ["interaction_id", "statement_order"]
).copy()

# Compute deltas WITHIN each interaction
groups = df_cmp.groupby("interaction_id")

for concept in concepts:
    df_cmp[f"delta_{concept}"] = (
        groups[f"cvp_{concept}"].diff()
    )

    df_cmp[f"abs_delta_{concept}"] = (
        df_cmp[f"delta_{concept}"].abs()
    )

# Relative progress through each interaction
df_cmp["step"] = groups.cumcount() + 1
df_cmp["n_steps"] = groups["step"].transform("size")

df_cmp["progress"] = (
    (df_cmp["step"] - 1) /
    (df_cmp["n_steps"] - 1).replace(0, 1)
)

print(df_cmp["defender_prompt_strategy"].value_counts())
```

```python
df_interaction = (
    df_cmp.groupby("interaction_id")
    .agg(
        condition=("defender_prompt_strategy", "first"),
        app=("app", "first"),
        n_statements=("step", "size"),
        believability=("cvp_believability", "mean"),
        enticingness=("cvp_enticingness", "mean"),
        delta_believability=("delta_believability", "mean"),
        delta_enticingness=("delta_enticingness", "mean"),
        abs_delta_believability=("abs_delta_believability", "mean"),
        abs_delta_enticingness=("abs_delta_enticingness", "mean"),
    )
    .reset_index()
)

# Summary table
display(
    df_interaction
    .groupby("condition")
    .agg(
        n=("interaction_id", "size"),
        believability=("believability", "mean"),
        enticingness=("enticingness", "mean"),
        delta_believability=("delta_believability", "mean"),
        delta_enticingness=("delta_enticingness", "mean"),
    )
    .round(3)
)
```

```python
metrics = [
    ("believability", "Believability"),
    ("enticingness", "Enticingness"),
    ("delta_believability", "Δ Believability"),
    ("delta_enticingness", "Δ Enticingness"),
]

conditions = ["vulnerable", "mislead"]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

for ax, (column, title) in zip(axes.flat, metrics):

    values = [
        df_interaction.loc[
            df_interaction["condition"] == condition,
            column
        ].dropna()
        for condition in conditions
    ]

    ax.boxplot(values)
    ax.set_xticks([1, 2], ["Vulnerable", "Mislead"])
    ax.set_title(title)
    ax.set_ylabel("Mean CVP score" if not column.startswith("delta") else "Mean Δ CVP")

    if column.startswith("delta"):
        ax.axhline(0, color="gray", linestyle="--")

    ax.grid(axis="y", alpha=0.2)

plt.suptitle("CVP comparison by defender condition")
plt.tight_layout()
plt.show()
```

## Explore how scores evolve over time

### dividing into phases (exploratory still)

```python
df_cmp["phase"] = pd.cut(
    df_cmp["progress"],
    bins=[-0.01, 1/3, 2/3, 1.01],
    labels=["Early", "Middle", "Late"]
)

# Calculate one average per interaction and phase
df_phases = (
    df_cmp.groupby(
        ["interaction_id", "defender_prompt_strategy", "phase"],
        observed=True
    )
    .agg(
        believability=("cvp_believability", "mean"),
        enticingness=("cvp_enticingness", "mean"),
        delta_believability=("delta_believability", "mean"),
        delta_enticingness=("delta_enticingness", "mean")
    )
    .reset_index()
)
```

```python
# check phase stacking - how many independent interactions contribute to each of the phases?
display(
    df_phases.groupby(
        ["defender_prompt_strategy", "phase"],
        observed=True
    )["interaction_id"]
    .nunique()
    .unstack()
)
```

```python
fig, axes = plt.subplots(2, 2, figsize=(13, 8))

metrics = [
    ("believability", "Believability"),
    ("enticingness", "Enticingness"),
    ("delta_believability", "Δ Believability"),
    ("delta_enticingness", "Δ Enticingness"),
]

phases = ["Early", "Middle", "Late"]
x = np.arange(len(phases))

for ax, (metric, title) in zip(axes.flat, metrics):

    for condition in ["vulnerable", "mislead"]:

        subset = df_phases.loc[
            df_phases["defender_prompt_strategy"] == condition
        ]

        stats = (
            subset.groupby("phase", observed=True)[metric]
            .agg(["mean", "sem"])
            .reindex(phases)
        )

        mean = stats["mean"].to_numpy(dtype=float)
        error = stats["sem"].fillna(0).to_numpy(dtype=float) * 1.96

        line, = ax.plot(
            x, mean, marker="o", label=condition.capitalize()
        )

        ax.fill_between(
            x, mean - error, mean + error,
            color=line.get_color(), alpha=0.15
        )

    if metric.startswith("delta"):
        ax.axhline(0, color="gray", linestyle="--")

    ax.set_xticks(x, phases)
    ax.set_title(title)
    ax.set_ylabel("CVP score")
    ax.set_xlabel("Interaction progress")
    ax.grid(alpha=0.2)
    ax.legend()

plt.suptitle("CVP trajectories by defender condition")
plt.tight_layout()
plt.show()
```

# Response transitions (Exchanges)

```python
def build_response_transitions(data, df_st):
    s = data["statements"]
    e = data["exchanges"]

    score_cols = ["cvp_believability", "cvp_enticingness"]
    scores = df_st[["statement_id", *score_cols]].drop_duplicates("statement_id")

    # 1. Locate the last defender COMMAND within each exchange
    commands = s.loc[
        s["actor"].eq("defender") & s["event"].eq("COMMAND"),
        ["exchange_id", "statement_id", "statement_order"]
    ]

    cmd_scores = commands.merge(scores, on="statement_id", how="left")

    response_info = cmd_scores.groupby("exchange_id", as_index=False).agg(
        response_order=("statement_order", "max"),
        n_defender_commands=("statement_id", "size"),
        defender_believability=("cvp_believability", "mean"),
        defender_enticingness=("cvp_enticingness", "mean")
    )

    # 2. Keep matched exchanges
    pairs = e.loc[
        e["complete"].eq(True)
        & e["trigger_attacker_statement_id"].notna(),
        ["interaction_id", "exchange_id", "exchange_order",
         "trigger_attacker_statement_id", "match_status"]
    ].copy()

    pairs = pairs.merge(
        response_info, on="exchange_id",
        how="inner", validate="one_to_one"
    )

    # 3. Recover triggering action and actual tool response
    trigger = s[
        ["interaction_id", "statement_id", "strategy",
         "action_body", "env_response", "statement_order"]
    ].rename(columns={
        "statement_id": "trigger_attacker_statement_id",
        "strategy": "strategy_before",
        "action_body": "action_before",
        "statement_order": "trigger_order"
    })

    pairs = pairs.merge(
        trigger,
        on=["interaction_id", "trigger_attacker_statement_id"],
        how="left",
        validate="many_to_one"
    )

    # 4. Find the FIRST attacker event AFTER the defender response
    following = s.loc[
        s["actor"].eq("attacker"),
        ["interaction_id", "statement_order", "statement_id",
         "event", "strategy", "action_body"]
    ].rename(columns={
        "statement_order": "next_order",
        "statement_id": "next_statement_id",
        "event": "next_event",
        "strategy": "strategy_after",
        "action_body": "next_action"
    })

    pairs = pd.merge_asof(
        pairs.sort_values("response_order"),
        following.sort_values("next_order"),
        by="interaction_id",
        left_on="response_order",
        right_on="next_order",
        direction="forward",
        allow_exact_matches=False
    )

    # 5. Attach CVP scores before and after the response
    before = scores.rename(columns={
        "statement_id": "trigger_attacker_statement_id",
        "cvp_believability": "belief_before",
        "cvp_enticingness": "entice_before"
    })

    after = scores.rename(columns={
        "statement_id": "next_statement_id",
        "cvp_believability": "belief_after",
        "cvp_enticingness": "entice_after"
    })

    pairs = pairs.merge(
        before, on="trigger_attacker_statement_id", how="left"
    ).merge(
        after, on="next_statement_id", how="left"
    )

    for concept, short in [
        ("believability", "belief"),
        ("enticingness", "entice")
    ]:
        pairs[f"delta_{concept}"] = (
            pairs[f"{short}_after"] - pairs[f"{short}_before"]
        )

    # 6. Add experimental condition and outcome
    pairs = pairs.merge(
        data["interactions"][
            ["interaction_id", "app", "defender_prompt_strategy",
             "agent_success"]
        ],
        on="interaction_id",
        how="left",
        validate="many_to_one"
    )

    return pairs


df_pairs = build_response_transitions(data, df_st)
```
